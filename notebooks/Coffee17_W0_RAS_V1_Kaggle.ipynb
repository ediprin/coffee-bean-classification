```python
# ONE-RUN KAGGLE PIPELINE: exact evidence -> exact Coffee17 -> 5-fold W0-RAS -> aggregate
CODE_COMMIT='2777f7f481cd2a357a5be3e90d1585561124689d'

import shutil, subprocess, sys
from pathlib import Path

INPUT=Path('/kaggle/input')
WORK=Path('/kaggle/working')
REPO=WORK/'coffee-bean-classification-w0ras'

assert INPUT.is_dir() and WORK.is_dir(), 'Notebook ini harus dijalankan di Kaggle.'

def run(command, cwd=None):
    print('\n$', ' '.join(map(str,command)), flush=True)
    subprocess.run([str(x) for x in command],cwd=cwd,check=True)

if REPO.exists():
    shutil.rmtree(REPO)

run([
    'git','clone','--quiet','--no-checkout',
    'https://github.com/ediprin/coffee-bean-classification.git',
    REPO,
])
run(['git','-C',REPO,'checkout','--quiet','--detach',CODE_COMMIT])

run([
    sys.executable,'-u',
    REPO/'scripts/kaggle_w0_ras_v1.py',
],cwd=REPO)

FINAL=WORK/'coffee17-w0-ras-v1/W0_RAS_V1_AGGREGATE.json'
ZIP=WORK/'coffee17-w0-ras-v1.zip'
assert FINAL.is_file(), FINAL
assert ZIP.is_file(), ZIP

print('\n=== COMPLETE ===')
print('AGGREGATE:',FINAL)
print('ZIP:',ZIP)
print('Gunakan Save Version agar checkpoint/output dapat dipakai untuk resume.')
```
