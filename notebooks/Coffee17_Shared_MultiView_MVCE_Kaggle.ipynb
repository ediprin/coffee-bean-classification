# Coffee17 Shared-Weight Multi-View — MVCE v1\n\n**Tidak membutuhkan saved primary output atau checkpoint teacher lama.**\n\nNotebook ini merekonstruksi split lama dari Coffee17 original, memverifikasi setiap validation fold terhadap authority yang dibekukan dari `kd-analysis-package.zip`, lalu menjalankan:\n\n- `R0_CONTROL`: kontrol baru pada environment yang sama;\n- `MVCE_ALL4`: satu MobileNetV3-Large shared untuk R0/C0/F0/W0 saat training; inference tetap R0 satu model.\n\nSebelum Run All: **cukup Add Input Coffee17 original**, aktifkan GPU dan Internet. Jangan attach output primary/KD lama.\n

```python
# Coffee17 shared-weight multi-view — matched R0 control + MVCE_ALL4
SCIENTIFIC_CODE_COMMIT = "d4d904838b98d60c39f06634b9c1991eae3b25e3"

import hashlib, importlib, json, os, shutil, subprocess, sys, zipfile
from pathlib import Path

INPUT = Path("/kaggle/input")
WORK = Path("/kaggle/working")
REPO = WORK / "coffee-bean-classification-shared-mv-code"
PROJECT = WORK / "coffee17-shared-multiview-project"

assert INPUT.is_dir() and WORK.is_dir(), "Notebook ini harus dijalankan di Kaggle."

def sha256_file(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

def merge_tree_exact(source, target):
    source, target = Path(source), Path(target)
    for item in sorted(source.rglob("*")):
        if not item.is_file():
            continue
        rel = item.relative_to(source)
        dst = target / rel
        dst.parent.mkdir(parents=True, exist_ok=True)
        if dst.is_file():
            if sha256_file(item) != sha256_file(dst):
                raise RuntimeError(f"Kaggle input conflict: {rel}")
        else:
            shutil.copy2(item, dst)

def run(command, cwd=None, log_path=None):
    command = [str(x) for x in command]
    print("\n$ " + " ".join(command), flush=True)
    if log_path is None:
        subprocess.run(command, cwd=cwd, check=True)
        return
    log_path = Path(log_path)
    log_path.parent.mkdir(parents=True, exist_ok=True)
    with log_path.open("a", encoding="utf-8") as stream:
        process = subprocess.Popen(
            command,
            cwd=cwd,
            stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT,
            text=True,
            bufsize=1,
        )
        assert process.stdout is not None
        for line in process.stdout:
            print(line, end="")
            stream.write(line)
            stream.flush()
        rc = process.wait()
    if rc:
        raise RuntimeError(f"Command gagal ({rc}): {' '.join(command)}")

# Optional resume from previous saved output of THIS notebook.
for prior in sorted(
    p for p in INPUT.rglob("coffee17-shared-multiview-project")
    if p.is_dir()
):
    print("MERGE PRIOR SHARED-MV OUTPUT:", prior)
    merge_tree_exact(prior, PROJECT)

PROJECT.mkdir(parents=True, exist_ok=True)
(PROJECT / "logs").mkdir(parents=True, exist_ok=True)
(PROJECT / "analysis").mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------------
# 1. Exact code checkout. No primary checkpoint/output is required.
# ------------------------------------------------------------------
if REPO.exists():
    shutil.rmtree(REPO)
run([
    "git", "clone", "--quiet", "--no-checkout",
    "https://github.com/ediprin/coffee-bean-classification.git", REPO
])
run(["git", "-C", REPO, "checkout", "--quiet", "--detach", SCIENTIFIC_CODE_COMMIT])

requirements = REPO / "requirements/preprocessing-study.txt"
run([sys.executable, "-m", "pip", "install", "-q", "-r", requirements])
run([sys.executable, "-m", "pip", "install", "-q", "--no-deps", "-e", REPO])
sys.path.insert(0, str(REPO / "src"))
importlib.invalidate_caches()
os.chdir(REPO)

import torch
if not torch.cuda.is_available():
    raise RuntimeError("Aktifkan Kaggle GPU sebelum Run All.")
print("GPU:", torch.cuda.get_device_name(0))

from bilinear_lmmd.data.preparation.prepare_coffee17 import discover_directory_samples
from bilinear_lmmd.data.preparation.audit_coffee17_provenance import audit_coffee17_provenance
from bilinear_lmmd.data.preparation.prepare_preprocessing_folds import prepare_preprocessing_folds
from bilinear_lmmd.data.preparation.materialize_preprocessing_development import (
    materialize_preprocessing_development,
)
from bilinear_lmmd.engine.shared_multiview import validation_identity_label_sha256

AUTHORITY = REPO / "configs/shared_multiview/validation_authority_v1.json"
R0_CONFIG = REPO / "configs/preprocessing_study/R0.yaml"
MVCE_CONFIG = REPO / "configs/shared_multiview/MVCE_ALL4.yaml"
OUT = PROJECT / "experiments/coffee17-shared-multiview-v1"
OUT.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------------
# 2. Reconstruct exact Coffee17 population and deterministic 5 folds.
# ------------------------------------------------------------------
print("\n=== 1/4 RECONSTRUCT COFFEE17 ===")
by_class = discover_directory_samples(INPUT)
raw_count = sum(len(v) for v in by_class.values())
print("Coffee17 mounted:", raw_count, "images /", len(by_class), "classes")
if raw_count != 979 or len(by_class) != 17:
    raise RuntimeError(
        "Input harus hanya memuat Coffee17 original yang valid "
        f"(expected 979/17, observed {raw_count}/{len(by_class)})."
    )

ARCHIVE = WORK / "coffee17_shared_mv_original.zip"
PROV_LOCAL = WORK / "coffee17_shared_mv_provenance"
CANONICAL = WORK / "coffee17_shared_mv_original_v1"
FOLDS_LOCAL = WORK / "coffee17_shared_mv_folds"

for path in (PROV_LOCAL, CANONICAL, FOLDS_LOCAL):
    shutil.rmtree(path, ignore_errors=True)
if ARCHIVE.exists():
    ARCHIVE.unlink()

with zipfile.ZipFile(ARCHIVE, "w", compression=zipfile.ZIP_STORED) as bundle:
    for class_name, paths in sorted(by_class.items()):
        for path in sorted(paths):
            info = zipfile.ZipInfo(
                f"{class_name}/{path.name}",
                date_time=(1980, 1, 1, 0, 0, 0),
            )
            info.compress_type = zipfile.ZIP_STORED
            info.external_attr = 0o644 << 16
            bundle.writestr(info, path.read_bytes())

provenance = audit_coffee17_provenance(
    ARCHIVE, PROV_LOCAL, canonical_root=CANONICAL
)
if provenance["decision"] != "PASS":
    raise RuntimeError(f"Provenance gagal: {provenance['decision']}")

fold_summary = prepare_preprocessing_folds(
    CANONICAL,
    PROV_LOCAL / "coffee17_provenance.json",
    FOLDS_LOCAL,
    folds=5,
    seed=42,
    validation_ratio=0.10,
)
if fold_summary["decision"] != "PASS_COFFEE17_PREPROCESSING_DATA_GATE":
    raise RuntimeError("Fold reconstruction gagal.")

authority = json.loads(AUTHORITY.read_text(encoding="utf-8"))
print("AUTHORITY SOURCE SHA:", authority["source_artifact_sha256"])

# ------------------------------------------------------------------
# 3. Matched R0 control + one new treatment (MVCE_ALL4).
# ------------------------------------------------------------------
print("\n=== 2/4 FIVE-FOLD MATCHED EXPERIMENT ===")

for fold in range(1, 6):
    print(f"\n================ FOLD {fold}/5 ================", flush=True)
    DEV = WORK / f"coffee17_shared_mv_dev_fold_{fold}"
    shutil.rmtree(DEV, ignore_errors=True)

    materialize_preprocessing_development(
        CANONICAL,
        FOLDS_LOCAL / "clean_manifest.json",
        FOLDS_LOCAL / "fold_manifest.json",
        DEV,
        fold=fold,
    )

    count, val_sha = validation_identity_label_sha256(DEV)
    expected = authority["folds"][f"fold_{fold}"]
    print("VAL COUNT:", count)
    print("VAL SHA  :", val_sha)
    if count != expected["count"] or val_sha != expected["identity_label_sha256"]:
        raise RuntimeError(
            f"Fold {fold}: validation reconstruction tidak cocok dengan authority."
        )
    print("VALIDATION AUTHORITY: PASS")

    run([
        sys.executable, "-u", "-m",
        "bilinear_lmmd.experiments.run_shared_multiview_control",
        "--config", R0_CONFIG,
        "--authority", AUTHORITY,
        "--fold", str(fold),
        "--data-root", DEV,
        "--output-root", OUT,
        "--required-commit", SCIENTIFIC_CODE_COMMIT,
        "--device", "cuda:0",
        "--resume",
        "--authorize-training",
    ], cwd=REPO, log_path=PROJECT / "logs" / f"R0_CONTROL_fold{fold}.log")

    run([
        sys.executable, "-u", "-m",
        "bilinear_lmmd.experiments.run_shared_multiview",
        "--config", MVCE_CONFIG,
        "--authority", AUTHORITY,
        "--fold", str(fold),
        "--data-root", DEV,
        "--output-root", OUT,
        "--required-commit", SCIENTIFIC_CODE_COMMIT,
        "--device", "cuda:0",
        "--resume",
        "--authorize-training",
    ], cwd=REPO, log_path=PROJECT / "logs" / f"MVCE_ALL4_fold{fold}.log")

    r0 = json.loads(
        (OUT / "R0_CONTROL" / f"fold_{fold}" / "seed42" / "result.json")
        .read_text(encoding="utf-8")
    )
    mv = json.loads(
        (OUT / "MVCE_ALL4" / f"fold_{fold}" / "seed42" / "result.json")
        .read_text(encoding="utf-8")
    )
    print(
        f"FOLD {fold} | "
        f"R0 Macro={r0['metrics']['macro_f1']:.4f} | "
        f"MVCE Macro={mv['metrics']['macro_f1']:.4f} | "
        f"Delta={mv['metrics']['macro_f1'] - r0['metrics']['macro_f1']:+.4f}"
    )

    shutil.rmtree(DEV, ignore_errors=True)
    torch.cuda.empty_cache()

# ------------------------------------------------------------------
# 4. Aggregate matched comparison and build a small analysis package.
# ------------------------------------------------------------------
print("\n=== 3/4 SUMMARY ===")
SUMMARY = PROJECT / "analysis" / "shared_multiview_summary.json"
run([
    sys.executable, "-u", "-m",
    "bilinear_lmmd.experiments.run_shared_multiview_summary",
    "--output-root", OUT,
    "--output", SUMMARY,
], cwd=REPO)

summary = json.loads(SUMMARY.read_text(encoding="utf-8"))
agg = summary["aggregate"]
print("\n=== MATCHED RESULT ===")
print(
    "R0 CONTROL Macro-F1 :",
    f"{agg['R0_CONTROL']['macro_f1']['mean']:.4%}"
)
print(
    "MVCE ALL4 Macro-F1  :",
    f"{agg['MVCE_ALL4']['macro_f1']['mean']:.4%}"
)
print(
    "Delta Macro-F1      :",
    f"{agg['MVCE_MINUS_MATCHED_R0']['macro_f1']['mean']:+.4%}"
)
print(
    "Delta Hard-F1       :",
    f"{agg['MVCE_MINUS_MATCHED_R0']['hard_class_f1']['mean']:+.4%}"
)
print(
    "Delta Worst-F1      :",
    f"{agg['MVCE_MINUS_MATCHED_R0']['worst_class_f1']['mean']:+.4%}"
)
print(
    "Historical init matched:",
    summary["all_historical_initial_matches"]
)

print("\n=== 4/4 BUILD ANALYSIS PACKAGE ===")
PKG = WORK / "shared-multiview-analysis-package"
shutil.rmtree(PKG, ignore_errors=True)
PKG.mkdir(parents=True)

shutil.copytree(PROJECT / "analysis", PKG / "analysis")
for method in ("R0_CONTROL", "MVCE_ALL4"):
    for fold in range(1, 6):
        src = OUT / method / f"fold_{fold}" / "seed42"
        dst = PKG / method / f"fold_{fold}"
        dst.mkdir(parents=True, exist_ok=True)
        for name in ("result.json", "run_contract.json", "history.json", "run_config.yaml"):
            p = src / name
            if p.is_file():
                shutil.copy2(p, dst / name)
        if (src / "validation").is_dir():
            shutil.copytree(src / "validation", dst / "validation")

shutil.copy2(AUTHORITY, PKG / "validation_authority_v1.json")
zip_path = Path(
    shutil.make_archive(
        str(WORK / "shared-multiview-analysis-package"),
        "zip",
        root_dir=PKG,
    )
)

print("\nREADY:", zip_path)
print("SIZE :", round(zip_path.stat().st_size / 1024 / 1024, 2), "MB")
print("PROJECT:", PROJECT)
print("Tidak ada outer-test inference dan tidak membutuhkan checkpoint primary lama.")

# Keep PROJECT + small analysis ZIP; remove temporary dataset/code copies.
for path in (REPO, PROV_LOCAL, CANONICAL, FOLDS_LOCAL, PKG):
    shutil.rmtree(path, ignore_errors=True)
if ARCHIVE.exists():
    ARCHIVE.unlink()
```
